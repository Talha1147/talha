```python
rice_price = 1800
rice_qty = 2

oil_price = 650
oil_qty = 3

sugar_price = 160
sugar_qty = 4

rice_cost = rice_price * rice_qty
oil_cost = oil_price * oil_qty
sugar_cost = sugar_price * sugar_qty

grand_total = rice_cost + oil_cost + sugar_cost

print("GROCERY STORE BILL")
print("RICE COST",rice_cost,"PKR")
print("OIL COST",oil_cost,"PKR")
print("SUGER COST",sugar_cost,"PKR")
print("GRAND TOTAL",grand_total,"PKR")
```

```python
basic_salary = 60000
house_allowance = basic_salary * 20 / 100
tax = basic_salary * 5 / 100
medical_allowance = basic_salary * 10 / 100

net_salary = basic_salary + house_allowance + medical_allowance - tax


print("EMPLOYEE SALARY DETAILS")
print("EMPLOYEE SALARY DETAILS")
print("Basic Salary:", basic_salary, "PKR")
print("House Allowance:", house_allowance, "PKR")
print("Medical Allowance:", medical_allowance, "PKR")
print("Tax:", tax, "PKR")
print("Net Salary:", net_salary, "PKR")
```

```python
distance = 360
average = 12
petrol_price = 275

petrol = distance / average
total_cost = petrol * petrol_price

print("FUEL EXPENSE")
print("Total Distance:", distance, "KM")
print("Petrol Required:", petrol, "Litres")
print("Petrol Price:", petrol_price, "PKR")
print("Total Expense:", total_cost, "PKR")
```

```python
length = 18
width = 12

area = length * width
perimeter = 2 * (length + width)

print("ROOM MEASUREMENTS")
print("Length:", length, "Feet")
print("Width:", width, "Feet")
print("Area:", area, "Square Feet")
print("Perimeter:", perimeter, "Feet")
```

```python
math = 78
english = 85
urdu = 90
physics = 72
chemistry = 80

obtained_marks = math + english + urdu + physics + chemistry

total_marks = 5 * 100

percentage = obtained_marks / total_marks * 100
print("STUDENT MARKSHEET")
print("Math:", math)
print("English:", english)
print("Urdu:", urdu)
print("Physics:", physics)
print("Chemistry:", chemistry)
print("Obtained Marks:", obtained_marks)
print("Total Marks:", total_marks)
print("Percentage:", percentage, "%")
```

```python
mobile_price = 240000
advance = 20000
months = 10

remaining = mobile_price - advance

monthly_installment = remaining / months
print("MOBILE INSTALLMENT DETAILS")
print("Mobile Price:", mobile_price, "PKR")
print("Advance Payment:", advance, "PKR")
print("Remaining Amount:", remaining, "PKR")
print("Total Months:", months)
print("Monthly Installment:", monthly_installment, "PKR")
```

```python
units = 320
rate = 38

cost = units * rate

tax = cost * 12 / 100

total_bill = cost + tax

print("ELECTRICITY BILL")
print("Units Used:", units)
print("Per Unit Rate:", rate, "PKR")
print("Units Cost:", cost, "PKR")
print("Tax Amount:", tax, "PKR")
print("Total Bill:", total_bill, "PKR")
```

```python
total_seconds = 7500

hours = total_seconds // 3600
remaining_seconds = total_seconds % 3600

minutes = remaining_seconds // 60
seconds = remaining_seconds % 60

print("Total Seconds:", total_seconds)
print("Hours:", hours)
print("Minutes:", minutes)
print("Remaining Seconds:", seconds)
```

```python
principal = 150000
rate = 8
time = 3

simple_interest = (principal * rate * time) / 100

total_amount = principal + simple_interest

print("Principal Amount:", principal, "PKR")
print("Annual Interest Rate:", rate, "%")
print("Time:", time, "Years")
print("Simple Interest:", simple_interest, "PKR")
print("Total Amount:", total_amount, "PKR")
```
